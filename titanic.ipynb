```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("titanic.csv")
```

```python
pclass = df["Pclass"].value_counts().sort_index()

plt.bar(["1st Class", "2nd Class", "3rd Class"], pclass)
plt.title("Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
gender = df["Sex"].value_counts()

plt.bar(gender.index, gender.values)
plt.title("Passengers by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("titanic.csv")
```

```python
survival = df["Survived"].value_counts()

plt.bar(["Did not survive", "Survived"], survival)
plt.title("Titanic Survival")
plt.xlabel("Result")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.pie(
    survival,
    labels=["Did not survive", "Survived"],
    autopct="%1.1f%%"
)
plt.title("Titanic Survival Percentage")
plt.show()
```

```python
age = df["Age"].dropna().sort_values().reset_index(drop=True)

plt.plot(age)
plt.title("Titanic Passenger Ages")
plt.xlabel("Passenger")
plt.ylabel("Age")
plt.show()
```
